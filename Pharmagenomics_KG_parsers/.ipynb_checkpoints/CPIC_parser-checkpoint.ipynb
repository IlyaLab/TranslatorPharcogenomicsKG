## Target edge properties to capture
* subject: MT-RNR1 (biolink:Gene)
* predicate: biolink:associated_with_response_to
* object: amikacin (biolink:ChemicalEntity/Drug)
* edge property:
    * FDA_label_category: (string enum: actionable PGx, informative PGx, testing required, testing recommended, no clinical PGx)
    * CPIC level (A/B/C/D)
        * A — there's a strong gene-drug association, and CPIC has published prescribing recommendations (genetic testing should influence the prescribing decision)
        * B — a gene-drug association exists but the evidence for changing prescribing is weaker, or CPIC hasn't yet issued a strong recommendation
        * C — few studies, weak/mixed evidence, unclear clinical action
        * D — little to no evidence of an association, or a genetic effect exists but no meaningful prescribing implication
    * ClinPGx level of evidence (1A/1B/2A/2B/3/4):
        * 1A — variant-specific prescribing guidance exists in a current clinical guideline or FDA label, backed by ≥1 supporting variant annotation
        * 1B — strong evidence (≥2 independent publications) but no variant-specific prescribing guidance yet
        * 2A — moderate evidence, and the gene is a Tier 1 Very Important Pharmacogene (VIP)
        * 2B — moderate evidence, replicated but with some inconsistency or small effect size
        * 3 — low evidence — single study, or multiple conflicting/non-significant studies
        * 4 — minimal evidence (case reports, in vitro only, etc.)

### Create the node dataframe first

```python
## now filtering to only include rows have values from drug_names_for_node_list or gene_names_for_node_list
mask = genes_nodes_sele_df['Symbol'].isin(gene_names_for_node_list)
nodes_genes_needed_df = genes_nodes_sele_df[mask][['NCBI Gene ID', 'Symbol']]
print(len(nodes_genes_needed_df))

mask = drugs_chemicals_nodes_sele_df['Name'].isin(drug_names_for_node_list) | drugs_chemicals_nodes_sele_df['Generic Names'].isin(drug_names_for_node_list) | drugs_chemicals_nodes_sele_df['Trade Names'].isin(drug_names_for_node_list)
nodes_drugs_needed_df = drugs_chemicals_nodes_sele_df[mask][['Name', 'PubChem Compound Identifiers']].drop_duplicates()
print(len(nodes_drugs_needed_df))
```

```python
### Node df need to have id, name, category as column names

## add category column first
nodes_genes_needed_df['category'] = 'biolink:Gene'
nodes_drugs_needed_df['category'] = 'biolink:Drug'

## then rename columns and make sure identifieres are in the correct form
nodes_genes_needed_df = nodes_genes_needed_df.rename(columns={'NCBI Gene ID': 'id', 'Symbol': 'name'})
nodes_genes_needed_df = nodes_genes_needed_df[['id', 'name', 'category']]

nodes_drugs_needed_df = nodes_drugs_needed_df.rename(columns={'PubChem Compound Identifiers': 'id', 'Name': 'name'})
nodes_drugs_needed_df = nodes_drugs_needed_df[['id', 'name', 'category']]

## convert idenitifiers into string to avoid issue when adding prefix
nodes_drugs_needed_df['id'] = nodes_drugs_needed_df['id'].astype(str)
nodes_genes_needed_df['id'] = nodes_genes_needed_df['id'].astype(str)

## add prefix as required by biolink
## add NCBIGene: prefix for genes and add PUBCHEM.COMPOUND: prefix for drugs
nodes_drugs_needed_df['id'] = 'PUBCHEM.COMPOUND:' + nodes_drugs_needed_df['id']
nodes_genes_needed_df['id'] = 'NCBIGene:' + nodes_genes_needed_df['id']
```

```python
## Filter out rows with null, no Clinical PGx values in col('PGx on FDA Label')
print("Before filter, row counts: ", len(edge_df))

## discussed with Guangrong to include those without lable
# edge_filter_df = edge_df.dropna(subset=['PGx on FDA Label'])
edge_filter_df = edge_df[~edge_df['PGx on FDA Label'].isin(['No Clinical PGx'])]

print("After filter, row counts: ", len(edge_filter_df))
```

```python
## Now rename to create neede columns
## subject: MT-RNR1 (biolink:Gene)
## predicate: biolink:associated_with_response_to
## object: amikacin (biolink:ChemicalEntity/Drug)
edge_filter_df = edge_filter_df.rename(columns={'Gene': 'subject_name', 'Drug': 'object_name'})

## add predicate column first
edge_filter_df['predicate'] = 'biolink:associated_with_response_to'

## then use left join on names to create new subject_id, subject_category, object_id, and object_category columns use col('id') and col('category') pandas dataframe node_final_df

## Merge for subject
edge_merged_df = edge_filter_df.merge(
    node_final_df[['name', 'id', 'category']],
    left_on='subject_name', right_on='name',
    how='left'
).rename(columns={'id': 'subject', 'category': 'subject_category'}).drop(columns='name')

## Merge for object
edge_merged_df = edge_merged_df.merge(
    node_final_df[['name', 'id', 'category']],
    left_on='object_name', right_on='name',
    how='left'
).rename(columns={'id': 'object', 'category': 'object_category'}).drop(columns='name')
```

```python
## Now handles the PMID column
edge_merged_df['publications'] = edge_merged_df['CPIC Publications (PMID)'].apply(
    lambda x: ' | '.join(f'PMID:{pmid.strip()}' for pmid in x.split(',')) if pd.notna(x) else x
)
```

```python
edge_parsing_df = edge_merged_df.copy()

### Put additional columns into attribute column
import pandas as pd
import json

ATTRIBUTE_CONFIG = {
    "PGx on FDA Label": {
        "attribute_type_id": "biolink:has_qualitative_value",
        "value_type_id": "your_namespace:FDA_PGx_label_category",
        "attribute_source": "infores:fda-label",
    },
    "CPIC Level": {
        "attribute_type_id": "biolink:has_qualitative_value",
        "value_type_id": "your_namespace:cpic_actionability_level",
        "attribute_source": "infores:cpic",
    },
    "ClinPGx Level of Evidence": {
        "attribute_type_id": "biolink:has_evidence",
        "value_type_id": "your_namespace:clinpgx_level_of_evidence",
        "attribute_source": "infores:clinpgx",
    },
}

def group_into_attributes(row, config):
    return [
        {
            "attribute_type_id": meta["attribute_type_id"],
            "value": str(row[col]).strip(),
            "value_type_id": meta["value_type_id"],
            "original_attribute_name": col,
            "attribute_source": meta["attribute_source"],
        }
        for col, meta in config.items()
        if pd.notna(row[col]) and str(row[col]).strip() != ""
    ]

edge_parsing_df["attributes"] = edge_parsing_df.apply(lambda r: group_into_attributes(r, ATTRIBUTE_CONFIG), axis=1)
```
